```python
import pandas as pd 
from sklearn.datasets import load_diabetes
from sklearn.preprocessing import StandardScaler
from xgboost import XGBRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, root_mean_squared_error
```

```python
d=load_diabetes()
df=pd.DataFrame(data=d.data,columns=d.feature_names)
df
```

```python
sc=StandardScaler()
df=sc.fit_transform(df)

df=pd.DataFrame(data=df,columns=d.feature_names)
df
```

```python
df['Target']=d.target
```

```python
x=df.drop(columns='Target')
y=df['Target']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
model=XGBRegressor()
df=model.fit(x_train,y_train)
model
```

```python
prediction = model.predict(x_test)
prediction
```

```python
print("Training Effiency :",model.score(x_train,y_train))
print("Testing Effiency :",model.score(x_test,y_test))
```

```python
mse=mean_squared_error(y_test,prediction)
mse
```

```python
rmse=root_mean_squared_error(y_test,prediction)
rmse
```

```python
mae=mean_absolute_error(y_test,prediction)
mae
```

```python
r=r2_score(y_test,prediction)
r
```
